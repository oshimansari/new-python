```python
# python program to find max and min in a list
```

```python
# Input list elements
numbers = list(map(int, input("Enter numbers seperated by space: ").split()))
#find maximum and minimum
maximum = max(numbers)
minimum = min(numbers)
print("Maximum value:", maximum)
print("Minimum value:", minimum)
```
