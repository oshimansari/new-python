```python
# python program to remove duplicates from  a list
```

```python
# Input list elements
numbers = list(map(int, input("Enter numbers seperated by space: ").split()))
# Remove duplicates
unique_list =[]
for num in numbers:
    if num not in unique_list:
        unique_list.append(num)
print("List after removing duplicates:")
print(unique_list)
```
